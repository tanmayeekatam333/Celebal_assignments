```python

from pyspark.sql.functions import *
from delta.tables import DeltaTable
df = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv("/Volumes/week7/default/week7/Sample - Superstore.csv")
)

display(df)
```

```python
print("Rows:", df.count())
print("Columns:", len(df.columns))

df.printSchema()
display(df.limit(10))
```

```python
df = df.fillna({
    "Postal Code": 0,
    "Sales": 0,
    "Quantity": 0,
    "Discount": 0,
    "Profit": 0
})
display(df.limit(10))

```

```python
rows_before = df.count()

df = df.dropDuplicates()

rows_after = df.count()

print("Before:", rows_before)
print("After:", rows_after)
```

```python
delta_path = "/Volumes/week7/default/week7/superstore_delta"

df.write.format("delta") \
.mode("overwrite") \
.save(delta_path)
from delta.tables import DeltaTable

delta_table = DeltaTable.forPath(
    spark,
    "/Volumes/week7/default/week7/superstore_delta"
)
```

```python
incremental_data = [
("CA-2017-152156","Second Class","CG-12520","Claire Gute","Consumer","United States","Kentucky","Henderson","42420","South","Furniture","Bookcases",300.00),
("CA-2017-138688","Second Class","DV-13045","Darrin Van Huff","Corporate","United States","California","Los Angeles","90036","West","Technology","Phones",500.00),
("NEW-10001","Standard Class","AB-10001","Amit Sharma","Consumer","India","Telangana","Hyderabad","500001","South","Technology","Laptops",950.00),
("NEW-10002","First Class","RS-10002","Ravi Kumar","Corporate","India","Karnataka","Bangalore","560001","South","Office Supplies","Binders",120.00)
]

columns = [
"Order ID","Ship Mode","Customer ID","Customer Name",
"Segment","Country","City","State","Postal Code",
"Region","Category","Sub-Category","Sales"
]

df_incremental = spark.createDataFrame(
    incremental_data,
    columns
)

display(df_incremental)
```

```python

delta_df = spark.read.format("delta").load(delta_path)
display(delta_df.limit(10))
```

```python
for col_name in df_incremental.columns:
    new_name = (
        col_name.replace(" ", "_")
                .replace("-", "_")
                .replace(".", "_")
    )
    df_incrementaldf_incremental = df_incremental.withColumnRenamed(
        col_name,
        new_name
    )

print(df_incremental.columns)
```

```python
from delta.tables import DeltaTable

delta_table = DeltaTable.forPath(spark, delta_path)

(
    delta_table.alias("target")
    .merge(
        df_incremental.alias("source"),
        "target.Order_ID = source.Order_ID"
    )
    .whenMatchedUpdate(set={
        "Sales": "source.Sales",
        "Customer_Name": "source.Customer_Name",
        "Ship_Mode": "source.Ship_Mode",
        "City": "source.City",
        "State": "source.State"
    })
    .execute()
)

print("MERGE completed successfully.")
```

```python
final_df = spark.read.format("delta").load(delta_path)

display(final_df)
```
