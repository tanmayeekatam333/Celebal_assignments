```python
df = spark.read.csv(
    "/Volumes/workspace/default/celebalweek6/sample_data.csv",
    header=True,
    inferSchema=True
)

display(df)
```

```python
df.filter(df.category == "Electronics") \
  .select("product_id", "price")
```

```python
from pyspark.sql.functions import col

df_new = df.withColumnRenamed("old_name", "new_name") \
           .withColumn("price",
                       col("price").cast("double"))
```

```python

df.filter(
    (col("status")=="Completed") &
    (col("amount")>1000)
).show()
```
